## **1. Comprensión del Negocio**

**Contexto financiero:** las entidades financieras enfrentan el desafío de evaluar
la capacidad de pago de solicitantes de crédito, muchos de los cuales tienen
historial crediticio limitado o inexistente. Otorgar crédito sin una evaluación
rigurosa expone a la entidad a pérdidas económicas significativas; ser
excesivamente restrictivo, en cambio, implica perder clientes solventes y
oportunidades comerciales.

**Necesidad del stakeholder:** el Área de Riesgo Financiero de la entidad, como
stakeholder de este proyecto, necesita anticipar antes de otorgar el crédito,
qué solicitantes tienen mayor probabilidad de incumplir sus pagos, para poder
ajustar las condiciones de aprobación (monto, tasa, garantías) o rechazar
operaciones de alto riesgo.

**Objetivo del proyecto:** desarrollar un modelo de Machine Learning (regresión
logística) que estime la probabilidad de incumplimiento (`TARGET`) de un
solicitante a partir de la información disponible al momento de la solicitud.
Puntualmente, esta primera evidencia busca responder:

>1. *¿Qué variables muestran patrones claramente distintos entre clientes que cumplen sus pagos y clientes que caen en default?*
2. *¿Qué problemas de calidad de datos deberán resolverse antes de poder construir un modelo predictivo confiable?*

En cuanto a una **solución basada en Machine Learning,** a diferencia de reglas fijas o la evaluación manual caso por caso, **un modelo entrenado con historial de miles de clientes permite identificar patrones no evidentes a simple vista,** aplicar el mismo criterio de forma consistente a gran escala, y priorizar la revisión humana en los casos de mayor riesgo. Esto es especialmente relevante en este dataset: con **122 variables** por solicitante, cruzar manualmente todas las combinaciones posibles de riesgo excede la capacidad de cualquier análisis caso por caso.

Dentro de un **Marco de Valor Esperado** podemos ver un desbalance de ~92% vs. ~8% en `TARGET` (se detalla en la sección 2.6)

Tenemos en cuenta que `application_train.csv` no es una muestra aleatoria de "toda la gente que solicitó crédito", sino de solicitudes que **ya pasaron el filtro de aprobación** de la entidad (scoring interno, políticas de riesgo vigentes). *El propio dataset refleja el criterio de riesgo que la entidad ya aplica.* A esto se suma que el incumplimiento es, en cualquier cartera crediticia real, un evento minoritario porque la mayoría de los tomadores de crédito paga sus deudas. Por eso creemos que el *modelo* deberá evaluarse con métricas que no ignoren a la clase minoritaria y que el umbral de decisión deberá ajustarse a ese desbalance, lo cual se aborda en la Evidencia 3.

>Un modelo que prediga "TARGET = 0" para el 100% de los solicitantes alcanzaría un **accuracy del 91.9%** sin haber aprendido nada y sería inútil para el negocio, porque fallaría en detectar exactamente al segmento que se quiere identificar.

Esto exige mirar la evaluación del modelo bajo un **Marco de Valor Esperado:** no todos los errores cuestan lo mismo.
- Un **Falso Negativo** (aprobar crédito a alguien que en realidad va a caer
  en default) implica una pérdida directa para la entidad: capital prestado
  que no se recupera.
- Un **Falso Positivo** (rechazar o poner condiciones más duras a un cliente
  que en realidad iba a cumplir) implica un costo de oportunidad: se pierde
  un cliente rentable.

En la industria crediticia, el costo de un Falso Negativo suele ser
sustancialmente mayor que el de un Falso Positivo, lo que va a orientar más
adelante la elección de métricas y el ajuste del umbral de decisión, no el accuracy.

### **2.3 Separación del Test Set (prevención de Data Snooping Bias)**

Antes de explorar relaciones entre variables o patrones de comportamiento,
apartamos un 20% de los datos como **Test Set en Cuarentena**. Esto evita
que nuestras decisiones de análisis y diseño del modelo queden
"contaminadas" por particularidades de la muestra.

- **`stratify=df['TARGET']`:** mantiene la misma proporción de clientes en
  default (~8%) tanto en train como en test.
- **`random_state=42`:** fija la semilla aleatoria para que la partición sea
  reproducible.

**Aclaración:** todo el EDA se realiza únicamente sobre **`train_df`**.

¿Por qué se construye un test propio y no se usa el de Kaggle? El desafío de Kaggle incluye un archivo de prueba (application_test.csv), pero ese archivo no contiene la variable TARGET: Kaggle la reserva para armar su ranking de competidores. Sin TARGET no se puede medir el desempeño del modelo (AUC, matriz de confusión, recall). Además, la consigna del ABP indica trabajar únicamente con application_train.csv. Por eso el conjunto de prueba se obtiene de ese mismo archivo.

Esta es la única división del proyecto. test_df queda aislado desde este momento y no participa del EDA, de la limpieza ni del entrenamiento. Solo se utiliza al final de la Evidencia 3 para evaluar los modelos. La comparación entre modelos durante el desarrollo se realiza con validación cruzada dentro de los datos de entrenamiento (sección 12), sin necesidad de volver a dividir los datos.



```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df['TARGET']
)

print(f"Train set: {train_df.shape}")
print(f"Test set: {test_df.shape}")
print()
print("Distribución TARGET en train:")
print(train_df['TARGET'].value_counts(normalize=True))
print()
print("Distribución TARGET en test:")
print(test_df['TARGET'].value_counts(normalize=True))
```

### **2.5 Análisis de Valores Faltantes**

Cuantificamos qué proporción de cada variable tiene **datos faltantes** y cruzamos
esa información con los bloques temáticos definidos en 2.4, para entender si
los faltantes se concentran en ciertos tipos de información (lo que sugeriría
una causa estructural) o están distribuidos de forma pareja entre bloques.


```python
#calculo de nulos y su porcentaje (pct)
nulos = train_df.isnull().sum()
nulos_pct = (train_df.isnull().mean() * 100).round(2)

#creamos la tabla resumen_nulos con los valores obtenidos
resumen_nulos = pd.DataFrame({'nulos': nulos, 'pct_nulos': nulos_pct})
resumen_nulos = resumen_nulos[resumen_nulos['nulos'] > 0].sort_values('pct_nulos', ascending=False)
# Ranking explícito (1° = variable con más nulos), para distinguir el orden
# aunque los porcentajes sean visualmente muy parecidos entre sí
resumen_nulos.insert(0, 'ranking', range(1, len(resumen_nulos) + 1))

print(f"Columnas con al menos 1 valor nulo: {len(resumen_nulos)} de {train_df.shape[1]}")
print()
print("Top 15 columnas con más nulos:")
resumen_nulos.head(15)
```

```python
#para saber a qué bloque pertenece cada columna con nulos
def obtener_bloque(col):
    for bloque, columnas in bloques_variables.items():   #recorremos cada clave:valor del diccionario de bloques
        if col in columnas:
            return bloque
    return "Sin clasificar"

#tomando cada nombre del índice de resumen_nulos y pasándolo por la función obtener_bloque
#Resultado final: resumen_nulos queda con una columna extra llamada 'BLOQUE'
resumen_nulos['bloque'] = resumen_nulos.index.map(obtener_bloque)

#Nulos totales por bloque
nulos_por_bloque = resumen_nulos.groupby('bloque').size().sort_values(ascending=False)
print("Cantidad de columnas con nulos, por bloque temático:")
print(nulos_por_bloque)
```

De las 67 columnas con al menos un valor faltante, **47
pertenecen al bloque de Características de la vivienda**  → el
100% de las variables de ese bloque tiene datos incompletos.
>Esto confirma la hipótesis planteada: la información del inmueble depende de que el solicitante la haya declarado (por ejemplo, no aplica si alquila), por lo que el faltante es de tipo **estructural**, no aleatorio.

El segundo bloque más afectado es *"Círculo social / consultas a buró"* (10 de 10 columnas), consistente con que **esa información depende de la existencia de registros externos.**

En contraste, los bloques "*Identificador/target*", "*Contacto/comunicación*" y *"Geográficas/regionales"* **no presentan ningún valor faltante,**  son datos que el sistema completa de forma obligatoria o automática al momento de
la solicitud.

*El tratamiento de estos valores faltantes se abordará en la segunda etapa del proyecto.*

```python
# Gráfico de barras top 15 variables con mayor % de nulos
# Todos los valores están entre 66% y 70%: el eje se acota al rango real de los datos
# y se agrega el ranking + porcentaje exacto sobre cada barra para que la diferencia sea legible.
plt.figure(figsize=(10, 6))
top15_nulos = resumen_nulos.head(15)

ax = sns.barplot(x=top15_nulos['pct_nulos'], y=top15_nulos.index,
                  hue=top15_nulos['bloque'], dodge=False)

minimo = top15_nulos['pct_nulos'].min()
maximo = top15_nulos['pct_nulos'].max()
plt.xlim(minimo - 1, maximo + 1)

for i, (valor, ranking) in enumerate(zip(top15_nulos['pct_nulos'], top15_nulos['ranking'])):
    ax.text(valor + 0.05, i, f'{ranking}° · {valor:.2f}%', va='center', fontsize=9)

plt.xlabel('% de valores faltantes')
plt.ylabel('Variable')
plt.title('Top 15 variables con mayor porcentaje de valores faltantes (Figura 1)')
plt.legend(title='Bloque temático', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### **3.1 Descripción de los Datos por Bloques**

Usamos los bloques temáticos que armamos en 2.4 para no tener que ir columna por columna. Nos centramos en los cuatro bloques que más nos interesan para el problema: **demográficas, laborales, financieras y scores externos.**

centinela sin limpiar (su máximo va a aparecer disparado), mientras que `YEARS_EMPLOYED` ya tiene ese valor tratado como nulo.


```python
def describir(cols, df=eda_df):
    """Estadísticos descriptivos + % de nulos de una lista de columnas numéricas."""
    cols = [c for c in cols if c in df.columns]
    num = df[cols].select_dtypes(include='number')
    tabla = num.describe().T
    tabla['nulos_%'] = (df[num.columns].isnull().mean() * 100).round(2)
    return tabla.round(2)


def describir_cat(cols, df=eda_df):
    """Para las categóricas: cuántas categorías tiene cada una y cuál domina."""
    cols = [c for c in cols if c in df.columns]
    cat = df[cols].select_dtypes(include='object')
    filas = []
    for c in cat.columns:
        vc = df[c].value_counts(normalize=True)
        filas.append({
            'variable': c,
            'categorías': df[c].nunique(),
            'más frecuente': vc.index[0],
            '% más frec.': round(vc.iloc[0] * 100, 1),
            'nulos_%': round(df[c].isnull().mean() * 100, 2)
        })
    return pd.DataFrame(filas)
```


### **3.3 Relación de las Variables con TARGET**

En esta sección se analiza qué variables permiten distinguir a los

El análisis consiste en comparar, para cada variable, cómo se comporta

Es importante aclarar que, dado el desbalance de clases observado en la

#### **3.3.1 Variables Numéricas**

Para cada variable numérica relevante, comparamos su **media, mediana y desvío estándar** entre los dos grupos de **`TARGET`**. Incluimos la mediana además de la media porque algunas de estas variables (como `AMT_INCOME_TOTAL`, cuyo máximo es 793 veces la mediana, como se vio en 3.2) son muy sensibles a valores extremos y en esos casos, la mediana es la lectura más confiable.

**Las variables `EXT_SOURCE_1`, `EXT_SOURCE_2` y `EXT_SOURCE_3`**

Refieren a **puntajes normalizados provenientes de fuentes de información
externas** a la entidad financiera, según indica la documentación oficial
del dataset. No se detalla públicamente qué mide cada fuente en
particular, pero conceptualmente funcionan de forma similar a un
**"puntaje de crédito"**: valores entre 0 y 1, donde un puntaje más alto
indica, en principio, menor riesgo asociado al solicitante.

Se incluyen tres variables (`EXT_SOURCE_1`, `2` y `3`) porque provienen
de tres fuentes distintas, cada una con su propio criterio de cálculo

```python
# Comparación de una variable entre los dos grupos de TARGET

def comparar_vs_target(col, df=eda_df):
    """Compara los estadísticos de una variable entre los dos grupos de TARGET."""
    tabla = df.groupby('TARGET')[col].agg(['count', 'mean', 'median', 'std']).round(3)
    tabla.index = ['TARGET=0 (cumple)', 'TARGET=1 (default)']
    return tabla

for col in ['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3',
            'YEARS_BIRTH', 'YEARS_EMPLOYED', 'AMT_INCOME_TOTAL', 'AMT_CREDIT']:
    print(f"--- {col} ---")
    display(comparar_vs_target(col))
```

En `AMT_INCOME_TOTAL` y `AMT_CREDIT`, en cambio, la diferencia de medianas entre ambos grupos es chica (por ejemplo, 148.500 vs. 135.000 en ingresos): los montos por sí solos no parecen separar bien a los dos grupos (retomamos

#### **3.3.2 Variables Categóricas**

Como ya explicamos en la introducción de esta sección, **comparamos la tasa de default de cada categoría (no la cantidad absoluta de defaults) contra la tasa general del 8.07%**, para no confundir "categoría grande" con "categoría riesgosa".

```python
# Tasa de Default por Categoría

def tasa_default(col, df=eda_df, min_clientes=100):
    """Tasa de default por categoría. Filtra las categorías con muy pocos casos."""
    tabla = (df.groupby(col, dropna=False)['TARGET']
               .agg(clientes='size', defaults='sum', tasa_default='mean'))
    tabla['tasa_default'] = (tabla['tasa_default'] * 100).round(2)
    tabla = tabla[tabla['clientes'] >= min_clientes]
    return tabla.sort_values('tasa_default', ascending=False)


TASA_GLOBAL = eda_df['TARGET'].mean() * 100
print(f"Tasa de default general: {TASA_GLOBAL:.2f}%\n")

for col in ['NAME_EDUCATION_TYPE', 'NAME_INCOME_TYPE', 'NAME_FAMILY_STATUS',
            'NAME_CONTRACT_TYPE', 'CODE_GENDER', 'FLAG_OWN_CAR']:
    print(f"--- {col} ---")
    display(tasa_default(col))
```

**El nivel educativo separa bastante bien a los dos grupos.**
- `Lower secondary` muestra una tasa de default de 11.07%, mientras que `Academic degree` muestra 1.61% → casi 7 veces de diferencia entre los extremos.

Vale aclarar que `Academic degree` tiene solo 124 clientes en
`train_df` (335 defaults en Lower secondary contra apenas 2 en Academic degree), así que aunque supera el filtro de `min_clientes` de la función, es una base bastante más chica que el resto de las categorías → la comparación es válida, pero con menos respaldo estadístico que las demás.

**El tipo de ingreso muestra el mismo tipo de gradiente.**
- `Working` tiene la tasa más alta (9.54%) y `Pensioner` la más baja (5.36%), con `Commercial associate` y `State servant` en un punto intermedio.
- Las categorías `Unemployed`, `Student`, `Businessman` y `Maternity leave`, visibles en la Figura 5 pero con muy pocos casos → no aparecen en esta tabla porque no superan el umbral mínimo de 100 clientes que usa la función `tasa_default()`.

De las otras cuatro variables categóricas analizadas, tres muestran diferencias que vale la pena registrar y una se mantiene cerca de la tasa general:

- `CODE_GENDER`: los hombres muestran una tasa de default más alta
  (10.14%) que las mujeres (7.00%).
- `NAME_CONTRACT_TYPE`: los créditos de tipo `Cash loans` tienen una tasa
  de default de 8.34%, más del doble que los `Revolving loans` (5.49%).
- `NAME_FAMILY_STATUS`: va de 5.87% (`Widow`) a 10.02% (`Civil marriage`), una diferencia mayor a la esperada dado que en 3.2 esta variable aparecía como la más balanceada del bloque en términos de cantidad de clientes por categoría → acá se ve que balance en composición no implica balance en riesgo.
- `FLAG_OWN_CAR` es la que menos separa: 8.48% (no tiene auto) contra

```python
# Buscamos pares de variables que digan casi lo mismo
corr_abs = eda_df[num_cols].corr().abs()

# Nos quedamos solo con la mitad de arriba de la matriz para no contar dos veces cada par
triangulo = corr_abs.where(np.triu(np.ones(corr_abs.shape), k=1).astype(bool))

pares = triangulo.stack().sort_values(ascending=False).reset_index()
pares.columns = ['variable_1', 'variable_2', 'correlación']

print(f"Pares con correlación mayor a 0.90: {(pares['correlación'] > 0.90).sum()}")
print("\nLos 15 más altos:")
display(pares.head(15).round(3))
```

La búsqueda sistemática encontró **63 pares de variables con correlación mayor a 0.90** → un número alto que confirma que buena parte del dataset contiene información duplicada.

Confirmamos el signo negativo aparte porque la búsqueda de pares usa el valor absoluto de la correlación.
> Tiene sentido de negocio: **sin empleo registrado, no hay teléfono laboral que declarar.** Para más adelante, esto significa que no tiene sentido incluir ambas variables en el modelo → aportarían exactamente la misma información.

El resto del top 15 confirma, a mayor escala, la sospecha que habíamosdejado pendiente en la Figura 10:

→ 13 de los 15 pares más correlacionados pertenecen al bloque de vivienda, entre sus variantes `_AVG` y `_MEDI`
(`YEARS_BUILD`, `FLOORSMIN`, `FLOORSMAX`, `ENTRANCES`, `ELEVATORS`, `COMMONAREA`, `LIVINGAREA`, `APARTMENTS`, `BASEMENTAREA`, `YEARS_BEGINEXPLUATATION`, `LIVINGAPARTMENTS`, `NONLIVINGAREA`,`LANDAREA`) → todas con** correlaciones entre 0.99 y 0.998.**
>Esto no es un problema aislado de una variable, sino un **patrón sistemático en todo el bloque** → consistente con lo que ya habíamos anticipado en 2.4: probablemente alcance con conservar una sola variante estadística por atributo de vivienda más adelante

```python
# Miramos si los tríos _AVG / _MODE / _MEDI del bloque de vivienda son redundantes
bases = ['APARTMENTS', 'YEARS_BEGINEXPLUATATION', 'FLOORSMAX', 'LIVINGAREA']

print("Correlación más baja dentro de cada trío:")
for b in bases:
    trio = [c for c in [f'{b}_AVG', f'{b}_MODE', f'{b}_MEDI']
            if c in eda_df.columns and eda_df[c].dtype != 'object']
    if len(trio) >= 2:
        m = eda_df[trio].corr()
        minima = m.where(~np.eye(len(m), dtype=bool)).min().min()
        print(f"  {b:<28} {minima:.3f}")
```

El chequeo confirma la sospecha para las cuatro bases probadas: en todos los casos, incluso el par menos correlacionado dentro de cada trío `_AVG` / `_MODE` / `_MEDI` **supera 0.96**
- `APARTMENTS`: 0.974
- `YEARS_BEGINEXPLUATATION`:0.967

### **5.7 Exclusión de la variable edad por criterio de equidad (Fairness)**

El EDA (sección 3.3.1) confirmó que 'YEARS_BIRTH' tiene una relación clara e inversa con el riesgo de incumplimiento:
>Los clientes más jóvenes presentan una tasa de default de hasta 2.5 veces la de los clientes de mayor edad.

Decidimos no utilizar la edad como variable predictora directa: esta decisión de "crédito justo" (fair lending) es usada en la práctica de la industria crediticia, donde las decisiones de otorgamiento deberían basarse en el mérito individual del
solicitante y no en atributos demográficos.

>Un modelo entrenado con esta variable podría penalizar sistemáticamente a clientes jóvenes solventes por pertenecer a ese grupo etario, en lugar de evaluarlos por su comportamiento financiero individual.

'YEARS_BIRTH'no se descarta del proyecto: **se separa en un conjunto aparte ('train_audit_fairness'),** excluida también de cualquier paso de imputación posterior (sección 5.8) para evitar que su influencia se filtre de forma indirecta a través de otra variable. Esta reserva es una preparación para la Evidencia 3, donde se plantea evaluar si el modelo, aun sin haber "visto" la edad durante el entrenamiento, termina produciendo predicciones sistemáticamente distintas entre clientes jóvenes y adultos.
La exclusión alcanza también a `DAYS_BIRTH`, la misma edad expresada en días (de la que se construyó `YEARS_BIRTH` en 5.1): conservarla habría dejado la edad dentro del modelo por otra vía.

```python
# Se reservan la edad y el género en un conjunto separado, exclusivamente para auditoría
# de equidad (Fairness) en la Evidencia 3. No se usan para entrenar ni para imputar.
train_audit_fairness = train_clean[['YEARS_BIRTH', 'CODE_GENDER']].copy()
train_clean = train_clean.drop(columns=['YEARS_BIRTH', 'DAYS_BIRTH', 'CODE_GENDER'])

print("YEARS_BIRTH, DAYS_BIRTH y CODE_GENDER removidas de train_clean y reservadas en train_audit_fairness "
      f"({train_audit_fairness.shape[0]:,} registros) para la auditoría de fairness en Evidencia 3.")
```

### **5.8 Imputación de EXT_SOURCE_1/2/3 con la mediana e indicadores de faltante**

Los tres scores externos son las variables con mayor correlación con `TARGET` de todo el dataset (sección 3.4), pero presentan distintos niveles de valores faltantes:
* 56.34% en `EXT_SOURCE_1`
* 19.84% en `EXT_SOURCE_3`
* 0.22% en `EXT_SOURCE_2`

*Dado su alto poder predictivo, se descarta la opción de eliminarlas.*

Se aplica el tratamiento que propone el material para los casos en que la ausencia del dato es informativa (Unidad II):

1. **Indicador de faltante (*missing flag*):** antes de imputar se crea una variable binaria por score (`EXT_SOURCE_1_FALTABA`, etc.) que vale 1 si el dato faltaba. El EDA mostró que la ausencia tiene relación con `TARGET` en `EXT_SOURCE_1` y `EXT_SOURCE_3` (sección 3.3.1), así que esa información se conserva.
2. **Imputación con la mediana:** los valores faltantes se completan con la mediana de cada score, calculada solo sobre `train_clean` (0.505, 0.566 y 0.535). Se usa la mediana porque, a diferencia de la media, no se ve afectada por valores extremos.

> Con estas dos piezas, el modelo dispone del valor imputado y, a la vez, sabe que ese valor no era real: puede asignarle un peso propio a "no tener el score".

Las medianas quedan guardadas (`medianas_ext`) para completar los faltantes de `test_df` con los mismos valores (sección 9). La edad no interviene en este paso, en línea con la decisión de la sección 5.7.

```python
### 5.8 Imputación de EXT_SOURCE_1/2/3 con la mediana + indicador de faltante

cols_ext = ['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']

# Missing flags: se crean ANTES de imputar, para conservar la información de que el dato faltaba
for col in cols_ext:
    train_clean[f'{col}_FALTABA'] = train_clean[col].isnull().astype(int)

# Medianas calculadas solo con train_clean: se guardan para completar igual a test_df (sección 9)
medianas_ext = train_clean[cols_ext].median()
train_clean[cols_ext] = train_clean[cols_ext].fillna(medianas_ext)

print("Medianas utilizadas para imputar:")
print(medianas_ext.round(4).to_string())
print(f"\nNulos restantes en los scores externos: {train_clean[cols_ext].isnull().sum().sum()}")
```

### **5.9 Nulos restantes en el bloque de vivienda**

Tras conservar únicamente la variante '_MEDI' de cada atributo de vivienda (sección 5.3), persisten valores faltantes en ese subconjunto de variables. Como se documentó en la sección 2.5, este patrón no es aleatorio:
>La información de vivienda depende de que el solicitante la haya declarado (por ejemplo, no aplica si alquila), por lo que el faltante es de tipo estructural.

Dado que las 14 variables del bloque comparten el mismo origen de faltante, no se justifica crear un indicador binario por cada una (lo que introduciría nuevamente redundancia, el mismo problema resuelto en 5.3). Se optó por un único indicador consolidado, 'SIN_DATO_VIVIENDA', que resume si el cliente presenta información de vivienda declarada o no. Los valores faltantes de las variables '_MEDI' se imputaron luego con la mediana de cada una, calculada sobre 'train_clean'.

```python
### 5.9 Bloque vivienda: las variables restantes son las _MEDI (ver 5.3)

cols_vivienda_medi = [c for c in train_clean.columns if c.endswith('_MEDI')]

# Un solo flag consolidado: alcanza con revisar si falta el primero de la lista,
# ya que el faltante es estructural y compartido por todo el bloque (ver 2.5)
train_clean['SIN_DATO_VIVIENDA'] = train_clean[cols_vivienda_medi[0]].isnull().astype(int)

print(f"SIN_DATO_VIVIENDA=1: {train_clean['SIN_DATO_VIVIENDA'].sum():,} "
      f"({train_clean['SIN_DATO_VIVIENDA'].mean()*100:.1f}%)")

# Imputación con mediana, columna por columna
for col in cols_vivienda_medi:
    train_clean[col] = train_clean[col].fillna(train_clean[col].median())

print(f"Nulos restantes en bloque vivienda: {train_clean[cols_vivienda_medi].isnull().sum().sum()}")
```

### **5.10 OCCUPATION_TYPE y OWN_CAR_AGE**

'OCCUPATION_TYPE presenta 31% de valores faltantes, correspondientes en su mayoría a clientes sin ocupación activa (jubilados, según se observó en la sección 3.3.2).
>Al tratarse de una variable categórica donde la ausencia del dato es en sí misma informativa y no un error, se optó por convertir los valores faltantes en una categoría explícita ("Sin_dato"), en lugar de imputarlos con la categoría más frecuente, lo que ocultaría esa distinción.

'OWN_CAR_AGE' presenta 66% de valores faltantes, un porcentaje que coincide con la proporción de clientes que no poseen vehículo ('FLAG_OWN_CAR = 'N'', ver sección 3.1). Dado que esa condición ya está representada explícitamente por 'FLAG_OWN_CAR', no se creó un indicador adicional: se consideró redundante. Los valores faltantes se completaron con 0, interpretándose como "no aplica antigüedad de vehículo".



```python


# OCCUPATION_TYPE: la ausencia es informativa, se conserva como categoría explícita
train_clean['OCCUPATION_TYPE'] = train_clean['OCCUPATION_TYPE'].fillna('Sin_dato')

# OWN_CAR_AGE: el faltante ya está representado por FLAG_OWN_CAR, se imputa con 0
train_clean['OWN_CAR_AGE'] = train_clean['OWN_CAR_AGE'].fillna(0)

print("Nulos en OCCUPATION_TYPE:", train_clean['OCCUPATION_TYPE'].isnull().sum())
print("Nulos en OWN_CAR_AGE:", train_clean['OWN_CAR_AGE'].isnull().sum())
```

### **5.11 Verificación final y tratamiento de nulos residuales**

Los pasos anteriores resolvieron los nulos de los bloques con mayor volumen o mayor
relevancia predictiva (vivienda, scores externos, ocupación). Antes de avanzar a la
Ingeniería de Características, se realiza una verificación final para confirmar que
no queden valores faltantes sin tratar en ninguna otra variable del dataset.

Se detectaron nulos residuales de bajo volumen en `AMT_ANNUITY` y `AMT_GOODS_PRICE`
(bloque financiero), y en las 10 variables del bloque "Círculo social / consultas a
buró" (`OBS_30/60_CNT_SOCIAL_CIRCLE`, `DEF_30/60_CNT_SOCIAL_CIRCLE`,
`AMT_REQ_CREDIT_BUREAU_*`). Estas variables se completan con la mediana de cada columna,
calculada sobre `train_clean`.
También quedan nulos en `YEARS_EMPLOYED` (clientes sin empleo registrado, que se mantienen como NaN desde la sección 5.1), en `NAME_TYPE_SUITE` (categórica) y en `DAYS_LAST_PHONE_CHANGE` (1 registro). `YEARS_EMPLOYED` se completa con 0, mismo criterio que `OWN_CAR_AGE` en 5.10: el faltante significa "no aplica" y esa información ya está en `SIN_EMPLEO_REGISTRADO`. `NAME_TYPE_SUITE` se completa con la categoría explícita 'Sin_dato', como `OCCUPATION_TYPE`.

```python
### 5.11 Verificación final y tratamiento de nulos residuales

# Nulos que quedan en train_clean después de 5.1 a 5.10
nulos_residuales = train_clean.isnull().sum()
nulos_residuales = nulos_residuales[nulos_residuales > 0]
print("Nulos residuales antes del tratamiento final:")
print(nulos_residuales)

# YEARS_EMPLOYED: se completa con 0 (no con la mediana), mismo criterio que
# OWN_CAR_AGE en 5.10: el faltante significa "no aplica" (cliente sin empleo
# registrado, ya capturado en SIN_EMPLEO_REGISTRADO), no un valor a estimar.
train_clean['YEARS_EMPLOYED'] = train_clean['YEARS_EMPLOYED'].fillna(0)

for col in nulos_residuales.index:
    if col == 'YEARS_EMPLOYED':
        continue  # ya tratada arriba
    if train_clean[col].dtype in ['float64', 'int64']:
        train_clean[col] = train_clean[col].fillna(train_clean[col].median())
    else:
        # Categóricas (ej. NAME_TYPE_SUITE): mismo criterio que OCCUPATION_TYPE (5.10)
        train_clean[col] = train_clean[col].fillna('Sin_dato')

print(f"\nNulos totales restantes en train_clean: {train_clean.isnull().sum().sum()}")
# Medianas de train_clean: se usan para completar los faltantes de test_df
# con valores aprendidos en el entrenamiento (sección 9)
medianas_train = train_clean.median(numeric_only=True)
```

### **6.1 y 6.2 Ratios de capacidad de pago**

'AMT_INCOME_TOTAL', 'AMT_CREDIT'y 'MT_ANNUITY', analizadas de forma aislada, mostraron una relación débil con 'TARGET' (sección 3.3.1). Sin embargo, el monto de un crédito o de una cuota solo tiene sentido de negocio en relación con el ingreso de quien lo solicita:
> Un mismo monto representa un riesgo muy distinto para un cliente con ingresos altos que para uno con ingresos bajos.

Se construyeron dos variables proporcionales (ratios):
- 'CUOTA_INGRESO_RATIO' = AMT_ANNUITY / AMT_INCOME_TOTAL → qué proporción del ingreso se destina a pagar la cuota del crédito.
- 'CREDITO_INGRESO_RATIO' = AMT_CREDIT / AMT_INCOME_TOTAL → qué múltiplo del ingreso anual representa el crédito solicitado.

Se espera que valores altos en ambos ratios estén asociados a mayor riesgo de incumplimiento, ya que representan un mayor compromiso del ingreso disponible.

Aclaración: los valores faltantes de AMT_ANNUITY (10 registros, ver sección 3.1) se completan en la sección 5.11, que se ejecuta antes de este paso; por eso CUOTA_INGRESO_RATIO no hereda ningún nulo.


```python
train_clean['CUOTA_INGRESO_RATIO'] = train_clean['AMT_ANNUITY'] / train_clean['AMT_INCOME_TOTAL']
train_clean['CREDITO_INGRESO_RATIO'] = train_clean['AMT_CREDIT'] / train_clean['AMT_INCOME_TOTAL']

print(train_clean[['CUOTA_INGRESO_RATIO', 'CREDITO_INGRESO_RATIO']].describe())
```

## **7. Transformación de Variables**

Con el dataset limpio y las nuevas variables construidas, resta adecuar la representación de los datos a los requerimientos del modelo que se utilizará en la Evidencia 3: una **regresión logística.**
>Este algoritmo asume relaciones lineales, es sensible a la escala de las variables numéricas y requiere que toda la información de entrada sea estrictamente numérica, lo cual condiciona las decisiones de esta sección.

### **7.1 Transformación logarítmica de variables monetarias**

En la sección 3.2 se identificó una asimetría marcada hacia la derecha en las variables monetarias, con el caso extremo de `AMT_INCOME_TOTAL` (máximo 793 veces la mediana). *Escalar variables no corrige por sí solo este tipo de distribución: solo cambia el rango de valores, no su forma.*
>Aplicar una transformación logarítmica antes de escalar reduce el peso de los valores altos y aproxima la distribución a una forma más simétrica, favoreciendo el desempeño de un modelo lineal como la regresión logística (Unidades II y III).

Se aplica `log(1 + x)` en lugar de `log(x)` para evitar valores indefinidos en caso de que alguna variable contenga ceros.

El efecto de la transformación se ve en la Figura 13: las cuatro variables pasan a tener una forma aproximadamente acampanada y **la media y la mediana quedan prácticamente superpuestas** (por ejemplo, 11.91 y 11.90 en `AMT_INCOME_TOTAL_LOG`). Cuando media y mediana coinciden, la distribución es aproximadamente simétrica, y el valor extremo de ingresos deja de dominar la escala.

```python
cols_monetarias = ['AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE']

for col in cols_monetarias:
    train_clean[f'{col}_LOG'] = np.log1p(train_clean[col])

# Histogramas después de la transformación, con la media y la mediana marcadas
fig, axes = plt.subplots(1, 4, figsize=(16, 3.8))
for ax, col in zip(axes, cols_monetarias):
    valores = train_clean[f'{col}_LOG']
    sns.histplot(valores, bins=50, color='#4C72B0', ax=ax)
    ax.axvline(valores.mean(), color='black', linestyle='--', linewidth=1, label='Media')
    ax.axvline(valores.median(), color='#C44E52', linestyle=':', linewidth=1.5, label='Mediana')
    ax.set_title(f'{col}_LOG', fontsize=10)
    ax.set_xlabel('')
    ax.set_ylabel('')
axes[0].legend()
fig.suptitle('Figura 13. Variables monetarias después de la transformación logarítmica', y=1.03)
plt.tight_layout()
plt.show()

# Media vs. mediana después del logaritmo: cuanto más se parecen, más simétrica es la distribución
media_mediana_log = pd.DataFrame({
    'media (log)': [train_clean[f'{c}_LOG'].mean() for c in cols_monetarias],
    'mediana (log)': [train_clean[f'{c}_LOG'].median() for c in cols_monetarias],
}, index=[f'{c}_LOG' for c in cols_monetarias])
print(media_mediana_log.round(2).to_string())
```

### **7.2 y 7.3 Codificación de variables categóricas**

Se utilizan dos técnicas distintas, según la naturaleza de cada variable (Unidad II):

**One-Hot Encoding** para las 9 variables categóricas nominales (sin orden natural entre categorías), como `NAME_INCOME_TYPE`, `NAME_FAMILY_STATUS` u `OCCUPATION_TYPE`. Esta técnica crea **una columna 0/1 por cada categoría** y evita introducir relaciones de orden inexistentes: asignar números arbitrarios (0, 1, 2…) haría que el modelo interprete, por ejemplo, que una categoría es "mayor" que otra sin que eso tenga sentido.

* Se usa `OneHotEncoder` de scikit-learn porque aprende las categorías una sola vez con `train_clean` y se guarda (`encoder_ohe`). Al aplicarlo a `test_df` devuelve siempre las mismas columnas, en el mismo orden.
* Las columnas de una misma variable suman siempre 1 (cada cliente pertenece a una sola categoría), es decir, son redundantes entre sí por construcción. La **regularización L2** de la regresión logística está pensada justamente para repartir el peso entre variables redundantes y estabilizar los coeficientes (Unidad III), por lo que se conservan todas las categorías. Por la misma razón, el coeficiente de una categoría se lee **comparándolo con las otras categorías de su misma variable** (secciones 10 y 13).
* Se guardan además las categorías presentes en `train_clean` y la **moda** (categoría más frecuente) de cada variable. Si en `test_df`, o en un cliente nuevo, aparece una categoría que no existe en el entrenamiento, se trata como un dato faltante y se completa con la moda, el criterio de imputación de la Unidad II para variables categóricas (sección 9).
* En `OCCUPATION_TYPE` (sección 5.10) y en `NAME_TYPE_SUITE` (sección 5.11), la categoría `Sin_dato` queda como columna propia del encoding, preservando la señal de "sin dato registrado".

El resultado son **59 columnas 0/1** generadas a partir de las 9 variables nominales.

**Ordinal Encoding** para `NAME_EDUCATION_TYPE`, la única variable categórica del análisis que sí posee un orden natural (de menor a mayor nivel educativo). En este caso, preservar el orden aporta información adicional que el One-Hot Encoding perdería.

Se evalúa también si `NAME_EDUCATION_TYPE_ORD` podría actuar como proxy de la edad excluida en 5.7: la correlación es débil (-0.148), por lo que se mantiene la variable sin riesgo relevante de filtración indirecta.

## **8. Revisión de multicolinealidad antes del modelado**

>La regresión logística es sensible a la multicolinealidad: cuando dos o más variables aportan la misma información, el modelo no puede repartir su efecto y los coeficientes se vuelven inestables y difíciles de interpretar.

Antes de entrenar el modelo baseline se revisan las variables.

1. **Correlación entre variables predictoras:** se buscan los pares de variables con correlación mayor a 0.90 y se grafica un mapa de calor (*heatmap*) de las variables candidatas.
2. **Redundancia por construcción:** variables que se calculan a partir de otras (por ejemplo, un promedio o un total) y que, por lo tanto, no aportan información nueva aunque su correlación individual no supere 0.90.

> Las columnas 0/1 de una misma variable categórica no se comparan entre sí: son complementarias por construcción (siempre suman 1, ver 7.2).

```python
# Correlación entre las variables predictoras
X_pred = train_clean.drop(columns=['TARGET'])
corr = X_pred.corr().abs()

#pares de variables distintas con correlación mayor a 0.90
pares = (corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
             .stack().reset_index())
pares.columns = ['variable_1', 'variable_2', 'correlacion']

#categoricas 0/1 suman 1 comp,ementarias.
origen = {f'{col}_{cat}': col for col, cats in zip(cols_nominales, encoder_ohe.categories_)
          for cat in cats}
misma_variable = (pares['variable_1'].map(origen).notna()
                  & (pares['variable_1'].map(origen) == pares['variable_2'].map(origen)))

pares_altos = pares[(pares['correlacion'] > 0.90) & ~misma_variable]
print("Pares de variables con correlación mayor a 0.90:")
print(pares_altos.sort_values('correlacion', ascending=False).round(3).to_string(index=False))

# heatmap de las variables candidatas a redundancia
vars_heatmap = ['CNT_CHILDREN', 'CNT_FAM_MEMBERS',
                'EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3', 'EXT_SOURCE_PROMEDIO',
                'OBS_30_CNT_SOCIAL_CIRCLE', 'OBS_60_CNT_SOCIAL_CIRCLE',
                'REGION_RATING_CLIENT', 'REGION_RATING_CLIENT_W_CITY',
                'AMT_INCOME_TOTAL_LOG', 'AMT_CREDIT_LOG', 'AMT_ANNUITY_LOG', 'AMT_GOODS_PRICE_LOG',
                'CUOTA_INGRESO_RATIO', 'CREDITO_INGRESO_RATIO']
plt.figure(figsize=(12, 9.5))
sns.heatmap(X_pred[vars_heatmap].corr(), annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, annot_kws={'size': 7})
plt.title('Figura 14. Correlación entre las variables predictoras candidatas a redundancia')
plt.tight_layout()
plt.show()
```

La revisión nos mostró cinco variables que sobraban, y las eliminamos:

* **`CNT_FAM_MEMBERS`** (miembros de la familia): es básicamente la cantidad de hijos (`CNT_CHILDREN`) más los adultos según el estado civil. Se construye con otras dos variables, así que no aporta nada nuevo (correlación de 0.88 con la cantidad de hijos).

* **`REGION_RATING_CLIENT`**: correlación de 0.951 con `REGION_RATING_CLIENT_W_CITY`. Nos quedamos con esta última, que en el EDA mostró más relación con `TARGET` (sección 3.4).

**Conjuntos de esta etapa**

1. **Conjunto de entrenamiento (`X_train`, `y_train`):** corresponde a `train_clean` completo (246.004 clientes y 126 variables). Se utiliza para que el modelo aprenda la relación entre las características y la variable objetivo (`TARGET`). **Origen del conjunto de evaluación:** el archivo de prueba de Kaggle (`application_test.csv`) no contiene la variable `TARGET`, y la consigna indica trabajar únicamente con `application_train.csv`. Por eso el conjunto de evaluación proviene de la división realizada en la sección 2.3, donde se detalla esta decisión.

**Muestreo estratificado:** el dataset está desbalanceado: como solo el 8% de los clientes cae en default, una división al azar podría dejar proporciones distintas en cada parte. Con `stratify` nos aseguramos de que los dos conjuntos tengan el mismo 8,07% de morosos.

**Preparación del conjunto de evaluación sin fuga de datos (*data leakage*):**  a `test_df` le aplicamos los mismos pasos de las secciones 5 a 8, en el mismo orden, con la función `preparar_test`, que cumple el rol del *Pipeline*. Todo lo que usa (medianas, categorías, modas, encoder y escalador) lo aprendimos solo de `train_clean`: sobre `test_df` solo *aplicamos* lo aprendido (`transform`), nunca *aprendemos* de él (`fit`).

En `test_df` no borramos ninguna fila, porque el modelo tiene que poder evaluar a cualquier cliente. Y si apareciera una categoría que no existe en el entrenamiento, se completa con la más frecuente (la moda, ver 7.2). En nuestro caso no aparece ninguna, pero la regla queda como resguardo para clientes nuevos.

```python
# ENTRENAMIENTO: todos los clientes de train_clean
X_train = train_clean.drop(columns=['TARGET'])
y_train = train_clean['TARGET']


def preparar_test(df):
    """Aplica a test_df los mismos pasos de las secciones 5 a 8,
    usando solo valores aprendidos en train_clean (sin data leakage)."""
    t = df.copy()

    # 5.1 Valor centinela de DAYS_EMPLOYED
    t['SIN_EMPLEO_REGISTRADO'] = (t['DAYS_EMPLOYED'] == CENTINELA).astype(int)
    t['YEARS_EMPLOYED'] = (-t['DAYS_EMPLOYED'] / 365.25).round(1)
    t.loc[t['SIN_EMPLEO_REGISTRADO'] == 1, 'YEARS_EMPLOYED'] = np.nan

    # 5.2, 5.3, 5.5 y 5.7: eliminación de las mismas columnas que en train
    t = t.drop(columns=['DAYS_EMPLOYED', 'FLAG_EMP_PHONE', 'ORGANIZATION_TYPE']
                       + cols_to_drop_present + casi_constantes
                       + ['SK_ID_CURR', 'DAYS_BIRTH', 'CODE_GENDER'])

    # 5.8 EXT_SOURCE: indicadores de faltante + medianas de train
    for col in cols_ext:
        t[f'{col}_FALTABA'] = t[col].isnull().astype(int)
    t[cols_ext] = t[cols_ext].fillna(medianas_ext)

    # 5.9 Indicador de vivienda (se calcula antes de completar los nulos)
    t['SIN_DATO_VIVIENDA'] = t[cols_vivienda_medi[0]].isnull().astype(int)

    # 5.10 y 5.11 Nulos: mismos criterios que en train
    t['OCCUPATION_TYPE'] = t['OCCUPATION_TYPE'].fillna('Sin_dato')
    t['OWN_CAR_AGE'] = t['OWN_CAR_AGE'].fillna(0)
    t['YEARS_EMPLOYED'] = t['YEARS_EMPLOYED'].fillna(0)
    t = t.fillna(medianas_train)                      # numéricas: medianas de train
    cols_texto = t.select_dtypes(include='object').columns
    t[cols_texto] = t[cols_texto].fillna('Sin_dato')  # categóricas

    # 6. Feature engineering
    t['CUOTA_INGRESO_RATIO'] = t['AMT_ANNUITY'] / t['AMT_INCOME_TOTAL']
    t['CREDITO_INGRESO_RATIO'] = t['AMT_CREDIT'] / t['AMT_INCOME_TOTAL']
    t['EXT_SOURCE_PROMEDIO'] = t[['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']].mean(axis=1)

    # 7.1 Logaritmos
    for col in cols_monetarias:
        t[f'{col}_LOG'] = np.log1p(t[col])

    # 7.2 y 7.3 Una categoría que no aparece en train se trata como dato faltante
    # y se completa con la moda de train; luego se codifica con el encoder ajustado en train
    for col in cols_nominales:
        t.loc[~t[col].isin(categorias_train[col]), col] = modas_train[col]
    t['NAME_EDUCATION_TYPE_ORD'] = t['NAME_EDUCATION_TYPE'].map(orden_educacion)
    ohe = pd.DataFrame(encoder_ohe.transform(t[cols_nominales]),
                       columns=encoder_ohe.get_feature_names_out(cols_nominales),
                       index=t.index)
    t = pd.concat([t.drop(columns=cols_nominales + ['NAME_EDUCATION_TYPE']), ohe], axis=1)

    # 7.4 Escalado con el scaler ajustado en train (solo transform, nunca fit)
    t = t.drop(columns=cols_monetarias_originales)
    t[cols_a_escalar] = scaler.transform(t[cols_a_escalar])

    # 8.1 Mismas eliminaciones por multicolinealidad
    t = t.drop(columns=list(cols_a_eliminar))
    return t


# EVALUACIÓN: test_df (el 20% separado en la sección 2.3)
test_prep = preparar_test(test_df)
X_test = test_prep[X_train.columns]   # mismas columnas y en el mismo orden que train
y_test = test_prep['TARGET']

print("--- CONJUNTOS DE ENTRENAMIENTO Y EVALUACIÓN ---")
print(f"Dimensiones de X_train: {X_train.shape}")
print(f"Dimensiones de X_test:  {X_test.shape}")
print(f"Proporción de TARGET=1 en Train: {y_train.mean():.4%}")
print(f"Proporción de TARGET=1 en Test:  {y_test.mean():.4%}")
print(f"Nulos en X_test: {X_test.isnull().sum().sum()}")
```

## **10. Entrenamiento del Modelo Baseline - Regresión Logística**

El modelo baseline es la referencia contra la que se mide cualquier mejora posterior. Se eligió una regresión logística debido a su alta interpretabilidad en el ámbito financiero, su eficiencia computacional y su compatibilidad estricta con las hipótesis planteadas en la etapa de preparación de datos.
> Este modelo devuelve una probabilidad de incumplimiento para cada cliente, y cada variable recibe un coeficiente con signo que indica si sube o baja el riesgo.

**En crédito eso importa, porque las decisiones tienen que poder explicarse.**

Configuración del baseline:

* **Desbalance de clases:** el modelo se entrena con las proporciones reales de los datos (92% de cumplidores y 8% en default).  Así, la probabilidad que da el modelo es el riesgo real del cliente: si dice 0.20, es un 20% de chances de default. El desbalance se maneja en la **decisión**, eligiendo el umbral (sección 11) en lugar de usar automáticamente el 0.50 por defecto.
* **`penalty='l2'` y `C=1.0`:** regularización L2 con la intensidad por defecto de scikit-learn.  Ayuda a que los coeficientes sean estables cuando hay variables relacionadas, como los montos y los ratios (8.3) o las categorías del One-Hot (7.2). Es un valor de partida; en la sección 12 lo optimizamos.
* **`solver='lbfgs'`:** es el método que usa scikit-learn para encontrar los coeficientes. Usamos el mismo en todos los modelos para que la comparación sea justa.
* **`max_iter=1000`:** le damos suficientes iteraciones para que llegue a la solución.
* **Datos:** el modelo se entrena con los 246.004 clientes de train_clean y se evalúa sobre test_df (sección 9).

>El modelo devuelve probabilidades y no una respuesta rígida de Sí o No.


Como el modelo devuelve probabilidades y no un "sí" o "no", el Área de Riesgo puede decidir a partir de qué probabilidad rechaza un crédito, buscando un equilibrio entre los préstamos que no se cobran (falsos negativos) y los buenos clientes que se pierden (falsos positivos). Eso lo analizamos en la sección 11

```python
#MODELO BASELINE
#Entrenamiento de Regresión Logística
from sklearn.linear_model import LogisticRegression

#Instanciamos el modelo Baseline
lr_baseline = LogisticRegression(
    penalty='l2',
    C=1.0,
    max_iter=1000,
    random_state=42,
    solver='lbfgs'
)

#Entrenar el modelo con los datos de entrenamiento
lr_baseline.fit(X_train, y_train)

#Obtener predicciones de probabilidad para la clase positiva (TARGET = 1)
y_pred_proba = lr_baseline.predict_proba(X_test)[:, 1]

print("✓ Modelo Baseline (Regresión Logística) entrenado exitosamente.")
```

```python
# Lectura de coeficientes en tres grupos:
# 1. Numéricas: están escaladas (media 0, desvío 1), por eso sus coeficientes se comparan entre sí.
# 2. Indicadores 0/1 (una sola columna): se leen como "tener la característica" frente a "no tenerla".
# 3. Categorías del One-Hot: las columnas de una misma variable suman siempre 1 (ver 7.2), por eso
#    su coeficiente se lee comparando categorías de esa misma variable entre sí.
# Solo se muestran columnas 0/1 donde ambos valores aparecen en al menos el 1% de los clientes:
# el coeficiente de una característica muy poco frecuente se estima con muy pocos casos.
def tabla_coeficientes(modelo, X, n=5):
    coef = pd.Series(modelo.coef_[0], index=X.columns)
    es_binaria = X.isin([0, 1]).all()
    frecuente = X.mean().between(0.01, 0.99)
    es_categoria = X.columns.isin(list(origen))

    for nombre, mascara in [('NUMÉRICAS (escaladas)', ~es_binaria),
                            ('INDICADORES 0/1', es_binaria & frecuente & ~es_categoria)]:
        c = coef[mascara].sort_values()
        print(f"\n===== Variables {nombre} =====")
        print(f"{n} que más REDUCEN la probabilidad de default:")
        display(c.head(n).round(3).to_frame('coeficiente'))
        print(f"{n} que más AUMENTAN la probabilidad de default:")
        display(c.tail(n)[::-1].round(3).to_frame('coeficiente'))

    # Categorías: dentro de cada variable, la de mayor y la de menor coeficiente
    filas = []
    cats = coef[es_categoria & frecuente]
    for variable in cols_nominales:
        c = cats[[col for col in cats.index if origen[col] == variable]]
        if len(c) >= 2:
            filas.append({'variable': variable,
                          'categoría de MAYOR riesgo': c.idxmax().replace(variable + '_', ''),
                          'categoría de MENOR riesgo': c.idxmin().replace(variable + '_', ''),
                          'diferencia de coeficientes': round(c.max() - c.min(), 3)})
    print("\n===== CATEGORÍAS (comparadas dentro de cada variable) =====")
    display(pd.DataFrame(filas).set_index('variable')
              .sort_values('diferencia de coeficientes', ascending=False))

tabla_coeficientes(lr_baseline, X_train)
```

#### **Lectura de los coeficientes del baseline**

Los coeficientes se leen en tres grupos, porque no todos son comparables entre sí:

* **Variables numéricas:** están escaladas (media 0, desvío 1), así que sus coeficientes se comparan directamente.
* **Indicadores 0/1:** se leen como "tener la característica" frente a "no tenerla".
* **Categorías del One-Hot Encoding:** se comparan con las otras categorías de su misma variable (7.2).

**Reducen el riesgo:** los puntajes externos son lo que más pesa (`EXT_SOURCE_3`, `2` y `1`, con coeficientes entre -0.21 y -0.49), como mostró el análisis exploratorio. También lo reducen más años de empleo, no tener empleo registrado (jubilados) y, entre las categorías, tener auto y ser pensionado/a.

**Aumentan el riesgo:** no tener los puntajes externos 1 y 3 (`EXT_SOURCE_1_FALTABA` y `_3_FALTABA`), una peor calificación de la región y, entre las ocupaciones, la de choferes (`Drivers`), que es la de mayor riesgo frente a la de contadores (`Accountants`).

**No se leen por separado:** `AMT_CREDIT_LOG` (+0.21) y `CREDITO_INGRESO_RATIO` (-0.19) tienen signos opuestos
> No significa que más endeudamiento reduzca el riesgo: es un efecto de la redundancia entre montos y ratios ( pasa porque una variable se calcula a partir de la otra (8.3), y el modelo reparte el efecto entre las dos), que se resuelve en la sección 12.2.

## **11. Evaluación del Modelo Baseline y Análisis de Umbrales**

En esta sección medimos qué tan bien funciona el baseline con el ROC-AUC y vemos cómo cambian los errores según el **umbral de decisión**, es decir, la probabilidad a partir de la cual consideramos que un cliente es riesgoso.

Probamos tres umbrales:

* **0.50**, el que usa scikit-learn por defecto;
* **0.08**, que es la tasa de default del entrenamiento: marca como riesgoso a quien tiene más riesgo que el promedio;
* **0.05**, más exigente, para detectar más morosos.

```python
#EVALUACIÓN DEL BASELINE Y COMPARACIÓN DE UMBRALES DE DECISIÓN

from sklearn.metrics import (
    roc_auc_score,
    classification_report,
    confusion_matrix
)
import matplotlib.pyplot as plt
import seaborn as sns

#ROC-AUC global
auc_score = roc_auc_score(y_test, y_pred_proba)
print(f"Score ROC-AUC del Modelo Baseline: {auc_score:.4f}\n")

#funcion para evaluar matriz de confusión y reporte de clasificación por umbral
def evaluar_modelo_umbral(y_true, y_probabilities, umbral=0.50):
    y_pred = (y_probabilities >= umbral).astype(int)

    print(f"=== REPORTE DE CLASIFICACIÓN (Umbral = {umbral}) ===")
    print(classification_report(y_true, y_pred, target_names=['Pagó (0)', 'Default (1)']))

    # grafico Matriz de Confusión
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=ax,
                xticklabels=['Pagó (0)', 'Default (1)'],
                yticklabels=['Pagó (0)', 'Default (1)'])
    plt.xlabel('Predicción del Modelo')
    plt.ylabel('Valor Real')
    plt.title(f'Matriz de Confusión - Umbral {umbral}')
    plt.tight_layout()
    plt.show()

#Umbrales de decisión
UMBRAL_DEFECTO = 0.50                      # valor por defecto de scikit-learn
UMBRAL_RIESGO = round(y_train.mean(), 2)   # tasa de default de entrenamiento: riesgo mayor al promedio
UMBRAL_ESTRICTO = 0.05                     # más exigente: detecta más morosos

print(f"Probabilidad promedio estimada por el modelo: {y_pred_proba.mean():.3f}")
print(f"Umbrales evaluados: {UMBRAL_DEFECTO} | {UMBRAL_RIESGO} | {UMBRAL_ESTRICTO}\n")

evaluar_modelo_umbral(y_test, y_pred_proba, umbral=UMBRAL_DEFECTO)
evaluar_modelo_umbral(y_test, y_pred_proba, umbral=UMBRAL_RIESGO)
evaluar_modelo_umbral(y_test, y_pred_proba, umbral=UMBRAL_ESTRICTO)
```

### **11.1 Qué mide el AUC-ROC**

La **curva ROC** muestra, para cada umbral posible, cuántos morosos detecta el modelo (eje vertical) y cuántos clientes buenos rechaza a cambio (eje horizontal).

El **AUC** es el área debajo de esa curva, y tiene una lectura muy simple: si tomamos al azar un cliente que no pagó y uno que sí, el AUC es la probabilidad de que el modelo le asigne más riesgo al que no pagó. Un AUC de 0.5 es lo mismo que ordenar a los clientes al azar, y un AUC de 1 sería ordenarlos a la perfección.

La elegimos como métrica principal por dos razones. Primero, no depende de ningún umbral: mide qué tan bien el modelo ordena a los clientes por riesgo, antes de decidir dónde cortar. Segundo, no se deja engañar por el desbalance. Un modelo que aprobara a todos tendría 91,93% de accuracy (porque acierta con todos los que pagan) y un AUC de 0.5. Algo parecido pasa con nuestro modelo usando el umbral 0.50: llega a 92% de accuracy, pero detecta menos del 1% de los morosos.



```python
from sklearn.metrics import roc_curve

fpr, tpr, _ = roc_curve(y_test, y_pred_proba)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f'Regresión logística (AUC = {auc_score:.4f})')
plt.plot([0, 1], [0, 1], '--', color='gray', label='Azar (AUC = 0.50)')
plt.xlabel('Cumplidores rechazados (tasa de falsos positivos)')
plt.ylabel('Morosos detectados (recall)')
plt.title('Curva ROC del modelo baseline')
plt.legend()
plt.tight_layout()
plt.show()
```

### **11.2 Análisis de umbrales**

Lo que el AUC no dice es cuántos clientes conviene rechazar. Eso lo vemos con el análisis de umbrales. Acá recorremos varios umbrales, entre 0.03 y 0.50, para ver todo el abanico: cuántos morosos detectamos y cuántos clientes buenos rechazamos en cada caso.
Como el modelo se entrena con la proporción real de morosos (8%), sus probabilidades son bajas (en promedio, 0.081), así que los umbrales que sirven están cerca de ese valor.

```python
filas = []
for u in [0.03, 0.04, 0.05, 0.06, 0.08, 0.10, 0.15, 0.20, 0.30, 0.50]:
    pred = (y_pred_proba >= u).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, pred).ravel()
    filas.append({
        'Umbral': round(u, 2),
        'Recall default': tp / (tp + fn),
        'Precision default': tp / (tp + fp) if (tp + fp) > 0 else np.nan,
        'F1 default': 2 * tp / (2 * tp + fp + fn),
        '% de clientes rechazados': (tp + fp) / len(y_test),
        'Morosos no detectados (FN)': fn,
        'Cumplidores rechazados (FP)': fp,
    })

tabla_umbrales = pd.DataFrame(filas).set_index('Umbral')
display(tabla_umbrales.style.format({
    'Recall default': '{:.1%}', 'Precision default': '{:.1%}',
    'F1 default': '{:.2f}', '% de clientes rechazados': '{:.1%}',
    'Morosos no detectados (FN)': '{:,.0f}', 'Cumplidores rechazados (FP)': '{:,.0f}'
}))
```

## **12. Desarrollo del Modelo ABP (Regresión Logística Optimizada)**

El modelo ABP parte del baseline y busca mejorarlo de dos formas que propone la consigna: **ajustando la regularización** (optimización de hiperparámetros) y **sacando variables que sobran** (selección de variables). Seguimos usando regresión logística, así la comparación con el baseline es directa.

**Para que la comparación sea justa**, todos los modelos de esta sección tienen lo mismo que el baseline:

* los mismos datos (los 246.004 clientes de `X_train`);
* la misma configuración (`solver='lbfgs'`, `penalty='l2'`, `max_iter=1000` y proporciones reales de default);
* la misma validación cruzada (5 pliegues estratificados).

Así, si el AUC cambia, sabemos que es por lo que estamos probando y no por otra cosa.

**¿Qué es la validación cruzada?** Dividimos `X_train` en 5 partes iguales, llamadas *pliegues*, y cada una mantiene el 8% de morosos. Entrenamos el modelo 5 veces: en cada vuelta aprende con 4 pliegues y se evalúa con el que quedó afuera. Al final miramos dos números:

* el **AUC medio**, que nos dice cuánto rinde el modelo en promedio;
* el **desvío estándar**, que nos dice qué tan estable es: si es chico, el resultado casi no depende de qué clientes tocaron en cada pliegue.

`test_df` no participa en ninguna de estas decisiones. Lo usamos una sola vez, en 12.3, para evaluar el modelo elegido.

> las medianas y el escalado de las secciones 5 a 8 se calcularon con todo `train_clean`, que incluye los cinco pliegues. Con 246.004 clientes, esos valores casi no cambian de un pliegue a otro, así que el efecto en la comparación es mínimo. En la evaluación final esto no pasa, porque `test_df` no participó de ningún cálculo.

### **12.1 Optimización del hiperparámetro C**

Un **hiperparámetro** es una configuración del modelo que no se aprende de los datos, sino que la elegimos nosotros antes de entrenar. En la regresión logística, el más importante es **C**, que controla cuánta regularización L2 se aplica. La regularización "castiga" los coeficientes muy grandes, y eso ayuda a que el modelo no se sobreajuste y a que los coeficientes de variables relacionadas sean estables. C es el inverso del λ (lambda) C = 1/λ.

Por eso C funciona al revés de lo que uno esperaría:

* **C chico** (por ejemplo, 0.001): regularización fuerte, coeficientes más cerca de cero, modelo más "conservador".
* **C grande** (por ejemplo, 1.0): regularización suave, coeficientes más libres.

Probamos cuatro valores (0.001, 0.01, 0.1 y 1.0) con **GridSearchCV**, que entrena y evalúa cada uno con la validación cruzada.

El valor 1.0 es el del baseline, así que el baseline también queda evaluado en la misma búsqueda. Nos quedamos con el C que da mayor AUC medio.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# Validaciom cruzada estratificada de 5 pliegues: la misma para todas las comparaciones
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

# Misma configuración que el baseline: lo único que cambia es C
modelo_lr = LogisticRegression(
    penalty='l2',
    solver='lbfgs',
    max_iter=1000,
    random_state=SEED
)

# C = 1.0 es el valor del baseline; los demás aplican una regularización más fuerte
param_grid = {'C': [0.001, 0.01, 0.1, 1.0]}

busqueda_c = GridSearchCV(modelo_lr, param_grid, cv=cv, scoring='roc_auc', n_jobs=2)
busqueda_c.fit(X_train, y_train)

tabla_c = pd.DataFrame({
    'C': param_grid['C'],
    'AUC medio': busqueda_c.cv_results_['mean_test_score'],
    'Desvío estándar': busqueda_c.cv_results_['std_test_score'],
}).set_index('C')
display(tabla_c.round(4))

mejor_c = busqueda_c.best_params_['C']
print(f"Mejor C según validación cruzada: {mejor_c}")
```

### **12.2 Selección de variables: montos redundantes**

En 8.3 mostramos que el monto del crédito (`AMT_CREDIT_LOG`) y el de la cuota (`AMT_ANNUITY_LOG`) repiten información de los ratios cuota/ingreso y crédito/ingreso, que se calculan a partir de ellos. Entonces probamos el modelo **sin esos dos montos**, con el mejor C de 12.1 y los mismos pliegues.

**La regla para decidir la definimos antes de ver los resultados:** si la diferencia de AUC entre los dos modelos es más chica que el desvío estándar entre pliegues, los consideramos equivalentes y nos quedamos con el que usa **menos variables**.

Esto es el principio de **parsimonia**: si dos modelos rinden parecido, conviene el más simple.

Después revisamos la correlación entre las variables financieras que quedan en el modelo.

```python
from sklearn.model_selection import cross_val_score

# Montos redundantes por construcción con los ratios de capacidad de pago (sección 8.2)
cols_montos = ['AMT_CREDIT_LOG', 'AMT_ANNUITY_LOG']

modelo_mejor_c = LogisticRegression(
    penalty='l2', solver='lbfgs', C=mejor_c,
    max_iter=1000, random_state=SEED
)

# AUC con todas las variables: resultado de la búsqueda de 12.1 para el mejor C
auc_con_media = busqueda_c.best_score_
auc_con_desvio = busqueda_c.cv_results_['std_test_score'][busqueda_c.best_index_]

# AUC sin los montos redundantes: mismos pliegues, mismo modelo
auc_sin = cross_val_score(modelo_mejor_c, X_train.drop(columns=cols_montos), y_train,
                          cv=cv, scoring='roc_auc', n_jobs=2)

tabla_vars = pd.DataFrame({
    'Variables': [X_train.shape[1], X_train.shape[1] - len(cols_montos)],
    'AUC medio': [auc_con_media, auc_sin.mean()],
    'Desvío estándar': [auc_con_desvio, auc_sin.std()],
}, index=['Con todas las variables', 'Sin montos redundantes'])
display(tabla_vars.round(4))

# Regla de parsimonia: si la diferencia de AUC es menor que el desvío entre pliegues,
# los dos modelos son equivalentes y se elige el que usa menos variables
diferencia = auc_con_media - auc_sin.mean()
if diferencia < auc_sin.std():
    vars_abp = X_train.drop(columns=cols_montos).columns
    print(f"Diferencia de AUC: {diferencia:.4f} → menor que el desvío ({auc_sin.std():.4f}): "
          "se elige el modelo sin montos redundantes.")
else:
    vars_abp = X_train.columns
    print(f"Diferencia de AUC: {diferencia:.4f} → mayor que el desvío ({auc_sin.std():.4f}): "
          "se conservan todas las variables.")
```

```python
# Correlación entre las variables financieras que quedan en el modelo ABP
vars_fin_abp = [c for c in ['AMT_INCOME_TOTAL_LOG', 'CUOTA_INGRESO_RATIO', 'CREDITO_INGRESO_RATIO']
                if c in vars_abp]
print(X_train[vars_fin_abp].corr().round(2).to_string())
```

### **12.3 Modelo ABP y evaluación final**

Entrenamos el modelo ABP con los 246.004 clientes de `X_train`, el mejor C (12.1) y las variables elegidas (12.2). Recién después lo evaluamos, una sola vez con `test_df` y comparamos su AUC con el del baseline (sección 11), medido sobre los mismos clientes. También calculamos el AUC con los datos de entrenamiento, para ver si hay sobreajuste.

```python
# MODELO ABP: mejor C (12.1) y variables elegidas (12.2), entrenado con todo X_train
X_train_abp = X_train[vars_abp]
X_test_abp = X_test[vars_abp]

lr_abp = LogisticRegression(
    penalty='l2', solver='lbfgs', C=mejor_c,
    max_iter=1000, random_state=SEED
)
lr_abp.fit(X_train_abp, y_train)

# Evaluación final sobre test_df (clientes que ningún modelo vio)
y_pred_proba_abp = lr_abp.predict_proba(X_test_abp)[:, 1]
auc_abp = roc_auc_score(y_test, y_pred_proba_abp)

print(f"Modelo ABP → C = {mejor_c} | variables: {X_train_abp.shape[1]}")
print(f"ROC-AUC en evaluación — Baseline: {auc_score:.4f} | ABP: {auc_abp:.4f}")

# Comparación entre entrenamiento y evaluación para detectar sobreajuste
auc_train_abp = roc_auc_score(y_train, lr_abp.predict_proba(X_train_abp)[:, 1])
print(f"ROC-AUC del ABP — entrenamiento: {auc_train_abp:.4f} | evaluación: {auc_abp:.4f}")
```

### **12.4 Interpretación de resultados**

**Sobre el valor de C (12.1):** los cuatro valores dan prácticamente el mismo AUC medio (entre 0.7426 y 0.7438). La diferencia entre el mejor C (0.1) y el del baseline (1.0) es menor a 0.0001, mucho más chica que el desvío entre pliegues (0.0024), así que en la práctica son iguales. Tiene sentido: con 246.004 clientes y 126 variables hay muchísimos datos por variable, así que el riesgo de sobreajuste es bajo y la regularización casi no cambia nada. El baseline ya estaba bien regularizado.

**Seleccion de variables (12.2):** sin `AMT_CREDIT_LOG` y `AMT_ANNUITY_LOG`, el AUC medio pasa de 0.7438 a 0.7435. Esa diferencia (0.0003) es menor que el desvío (0.0022), así que, siguiendo la regla de parsimonia, elegimos el modelo sin montos: **124 variables**. Las variables financieras que quedan (el ingreso y los dos ratios) ya no se calculan unas a partir de otras, así que sus coeficientes se pueden leer por separado (sección 13).

**Evaluación final (12.3):** con los 61.503 clientes de `test_df`, el ABP obtiene un AUC de **0.7457** y el baseline, **0.7464**. La diferencia (0.0007) es menor que la variación normal entre pliegues, así que los dos modelos predicen igual de bien. Además, el AUC del ABP es casi el mismo en entrenamiento (0.7450), en validación cruzada (0.7435) y en prueba (0.7457). Según la Unidad V, cuando el error de entrenamiento y el de prueba son parecidos, el modelo **generaliza**: le va igual de bien con clientes que no conoce, y no hay señales de sobreajuste.

**Entonces, ¿el ABP mejora al baseline?** En AUC no, pero sí en **simplicidad e interpretabilidad**: predice lo mismo con dos variables menos y sin montos repetidos, así que cada coeficiente se puede explicar por separado ante el Área de Riesgo o un organismo de control. Que ninguna configuración cambie el AUC nos dice algo importante: el límite no está en cómo ajustamos el modelo, sino en la información que hay en `application_train.csv`. Para predecir mejor haría falta más información de los clientes, como su historial en otras entidades.

**Los documentos presentados** (`FLAG_DOCUMENT_3`, `6` y `16`) aparecen asociados al riesgo, pero no les encontramos una explicación de negocio clara. Son asociaciones, no causas.

En este cuadro comparamos el baseline y el ABP, evaluados con los mismos 61.503 clientes de `test_df` (que ninguno usó para entrenar) y con los mismos umbrales (0.08 y 0.05, sección 11).

**Criterio de selección** buscamos el mejor equilibrio entre detectar morosos (recall) y no rechazar de más a los buenos clientes (precisión y % de rechazados). Si los dos rinden parecido, preferimos el más simple.

```python
from sklearn.metrics import recall_score, precision_score, f1_score

modelos = {
    'Baseline: Regresión Logística': y_pred_proba,
    'Modelo Final (ABP): Logística Optimizada': y_pred_proba_abp,
}

filas = []
for nombre, proba in modelos.items():
    for umbral in [UMBRAL_RIESGO, UMBRAL_ESTRICTO]:
        pred = (proba >= umbral).astype(int)
        filas.append({
            'Modelo': nombre,
            'Umbral': umbral,
            'ROC-AUC': round(roc_auc_score(y_test, proba), 4),
            'Recall Default': f"{recall_score(y_test, pred):.1%}",
            'Precision Default': f"{precision_score(y_test, pred):.1%}",
            'F1 Default': round(f1_score(y_test, pred), 3),
            '% Rechazados': f"{pred.mean():.1%}",
        })

df_comparativo = pd.DataFrame(filas)
print("=" * 78)
print("CUADRO COMPARATIVO DE DESEMPEÑO DE MODELOS")
print("=" * 78)
display(df_comparativo)
```

* **Con 0.05:** detecta el 84.7%, pero rechaza al 55.7% y la precisión baja al 12.3%.

Qué umbral usar depende de los costos reales de la entidad: cuánto se pierde con un crédito impago frente a cuánto se deja de ganar al rechazar a un buen cliente.

```python
# Edad y género de los clientes de evaluación: se toman de test_df solo para la auditoría
aud = test_df.loc[X_test.index, ['CODE_GENDER']].copy()
aud['YEARS_BIRTH'] = (-test_df.loc[X_test.index, 'DAYS_BIRTH'] / 365.25).round(1)
aud['y'] = y_test.values
aud['proba'] = y_pred_proba_abp
aud['rechazado'] = (aud['proba'] >= UMBRAL_RIESGO).astype(int)
aud['Grupo de edad'] = pd.cut(aud['YEARS_BIRTH'], [0, 30, 40, 50, 60, 120])

def resumen(g):
    return pd.Series({
        'Casos': len(g),
        'Default real': f"{g['y'].mean():.1%}",
        'AUC': round(roc_auc_score(g['y'], g['proba']), 3),
        'Recall': f"{g.loc[g['y'] == 1, 'rechazado'].mean():.1%}",
        '% Rechazados': f"{g['rechazado'].mean():.1%}",
    })

# 'XNA' (género sin dato) tiene muy pocos casos y no permite calcular un AUC
print("Por género:")
display(aud[aud['CODE_GENDER'] != 'XNA'].groupby('CODE_GENDER')[['y', 'proba', 'rechazado']].apply(resumen))

print("\nPor grupo de edad:")
display(aud.groupby('Grupo de edad', observed=True)[['y', 'proba', 'rechazado']].apply(resumen))
```

### **Revisión de equidad por género y edad**

Se evalúa el modelo ABP (umbral 0.08) sobre los clientes de evaluación (`test_df`), separando por género y por grupo de edad, con las mismas métricas de la sección 11 (AUC, recall y % de rechazados). Estas dos variables no forman parte del modelo: se toman de `test_df` solo para esta revisión, en línea con las restricciones éticas.

> Los hombres tienen más incumplimiento real (10.2% vs 7.0%) y el modelo los rechaza más (42.2% vs 31.5%), en proporción a ese mayor riesgo.

**Edad:** aunque el modelo no usa la edad, sus resultados cambian mucho según el grupo.
> Se rechaza al 59.3% de los menores de 30 años y al 18.1% de los mayores de 60: más de tres veces más. Esa diferencia es mayor que la del incumplimiento real (11.3% vs 5.0%, algo más del doble).

> En los mayores de 60, además, el modelo ordena peor (AUC 0.677) y detecta solo el 40% de los morosos.

**Conclusión:** excluir la edad no alcanza para evitar su efecto, porque otras variables relacionadas con ella (como los años de empleo o ser pensionado/a) la reflejan de forma indirecta.

## **Cierre del proyecto**

**Pregunta de negocio:**
*¿Cómo identificar anticipadamente a los clientes con mayor probabilidad de incumplir un crédito?*

>Con un modelo de regresión logística optimizado sobre datos históricos, que ordena a los clientes según su probabilidad de default con un ROC-AUC de **0.7457**, medido sobre clientes que el modelo no vio durante su desarrollo.

Esto significa que, ante un cliente moroso y uno pagador elegidos al azar, el modelo asigna mayor riesgo al moroso en 75 de cada 100 casos.

*Es un poder de predicción moderado:* sirve como scorecard de apoyo para automatizar y priorizar la evaluación de crédito, no para tomar decisiones sin supervisión.

>**Síntesis del proceso (CRISP-DM):**

Siguiendo la metodología CRISP-DM, se separa al inicio un 20% de los clientes para la evaluación final. Sobre el 80% restante se depuran los datos: control de duplicados, tratamiento del dato centinela `365243` en antigüedad laboral, imputación con la mediana e indicadores de faltante, transformación logarítmica de los montos y eliminación de variables constantes y redundantes.

También se construyen ratios clave como la relación cuota/ingreso, se excluyen género y edad del modelo y se trata el desbalance de clases (solo ~8% de morosos) con la elección del umbral de decisión.

Los modelos se comparan con validación cruzada de 5 pliegues y se evalúan una sola vez sobre el 20% reservado.

>**Principales hallazgos:**

* **Factores determinantes:** los puntajes externos (`EXT_SOURCE_1`, `2` y `3`) son los predictores de mayor peso. Una cuota elevada respecto del ingreso aumenta el riesgo; la estabilidad laboral y el nivel educativo lo reducen.
* **Optimización y selección de variables:** ajustar la regularización (C = 0.1) y quitar los dos montos redundantes no cambia el AUC (0.7457 vs. 0.7464 del baseline), pero elimina la redundancia y deja un modelo más simple y explicable.
* **Límite de la información:** que ninguna configuración del modelo cambie el AUC indica que el techo predictivo está en la información disponible en `application_train.csv`, no en el ajuste del modelo.
* **Impacto de umbrales:** cambiar el umbral no mejora el modelo, solo redistribuye los errores. Con el umbral por defecto (0.50) el modelo es inútil para el negocio; al pasar de 0.08 a 0.05, la detección de morosos sube del 68.6% al 84.7%, pero el rechazo total sube del 35.2% al 55.7%.

>**Limitaciones del análisis:**

* **Precisión en rechazos:** con el umbral 0.08, aproximadamente 1 de cada 6 clientes rechazados es un moroso real (precisión del 15.8%), lo que implica un costo de oportunidad por clientes aptos descartados.
* **Sesgo indirecto por edad:** aunque el modelo no utiliza la edad, la revisión de equidad muestra que rechaza con mayor frecuencia a los menores de 30 años (59.3%) y que pierde capacidad de discriminación en los mayores de 60 (AUC 0.677), por su relación con variables como la antigüedad laboral.
* **Información disponible:** el modelo utiliza solo los datos de la solicitud (`application_train.csv`). No incluye el historial crediticio del cliente en otras entidades ni su comportamiento de pago en créditos anteriores.
* **Validación en un único período:** los clientes de evaluación pertenecen al mismo período que los de entrenamiento, por lo que no se mide cómo se comporta el modelo si cambian las condiciones económicas.

> **Recomendaciones para la implementación en producción:**

* **Calibración del umbral económico:** definir el punto de corte definitivo con la matriz de costos real de la entidad (pérdida por default vs. margen financiero no percibido).
* **Auditoría de equidad y riesgo legal:** evaluar con las áreas normativa y legal los sesgos indirectos por edad antes de desplegar el modelo.
* **Monitoreo continuo:** validar el modelo con datos de un período posterior (*Out-of-Time*) y monitorear si cambian los datos de los solicitantes (*Data Drift*).
